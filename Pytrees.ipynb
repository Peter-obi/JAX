```python
import jax
import jax.numpy as jnp
from jax import random
```

```python
LAYER_SIZES =[200 * 200 * 3, 2048, 1024, 2]
PARAM_SCALE = 0.01
```

```python
def random_layer_params(m, n, key, scale=1e-2):
  w_key, b_key = random.split(key)
  return(scale * random.normal(w_key, (n, m))), (scale * random.normal(b_key, (n,)))
```

```python
def init_network_params(sizes, key=random.PRNGKey(0), scale=0.01):
  keys = random.split(keys, len(sizes) - 1)
  return [random_layer_params(m, n, k, scale) for m, n, k in zip(sizes[:-1], sizes[1:], keys)] #given sizes initialize all layers of a FNN
```
